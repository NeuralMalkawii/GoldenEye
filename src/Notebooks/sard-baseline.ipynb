```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 1.4 — Random Seed Fixation
# ─────────────────────────────────────────────────────────────────────────────
# Fixing seeds across Python, NumPy, and PyTorch ensures that:
#   • Data shuffling is the same every run
#   • Weight initialization is identical
#   • Dropout patterns are reproducible
# Note: torch.backends.cudnn.deterministic = True can hurt speed;
#       it's optional and commented out by default.

SEED = 42   # Change this to any integer to get a different-but-reproducible run

def set_seeds(seed: int = 42) -> None:
    """Fix random seeds for reproducibility across all frameworks."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)   # For multi-GPU setups
    os.environ["PYTHONHASHSEED"] = str(seed)
    # torch.backends.cudnn.deterministic = True  # Uncomment for full determinism
    # torch.backends.cudnn.benchmark = False      # Uncomment for full determinism
    print(f"✅ Random seeds fixed to {seed}")

set_seeds(SEED)
import subprocess, sys

def pip_install(package: str) -> None:
    """Install a pip package quietly, printing only errors."""
    result = subprocess.run(
        [sys.executable, "-m", "pip", "install", "-q", package],
        capture_output=True, text=True
    )
    if result.returncode != 0:
        print(f"❌ Failed to install {package}:\n{result.stderr}")
    else:
        print(f"✅ {package} installed successfully")

pip_install("ultralytics>=8.2.0")   # Core YOLOv8 framework
pip_install("albumentations")        # Advanced augmentation library
pip_install("opencv-python-headless") # Computer vision utilities
pip_install("seaborn")               # Statistical visualization
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 1.5 — Path Configuration
# ─────────────────────────────────────────────────────────────────────────────
# ✏️  EDIT THIS CELL — update DATASET_ROOT to point to your Kaggle dataset.
#
# When you add a dataset to a Kaggle notebook it is mounted at:
#   /kaggle/input/<dataset-slug>/
# You can find the exact path in the right panel under "Data".
#
# Expected dataset layout (YOLO format):
#   <DATASET_ROOT>/
#       images/
#           train/   ← training images
#           val/     ← validation images
#           test/    ← (optional) test images
#       labels/
#           train/   ← .txt label files for training images
#           val/     ← .txt label files for validation images
#           test/    ← (optional)
#       data.yaml    ← (optional, we will generate one if missing)

# ── Paths ──────────────────────────────────────────────────────────────────
DATASET_ROOT  = Path("/kaggle/input/datasets/nikolasgegenava/sard-search-and-rescue/search-and-rescue")   # ✏️ CHANGE THIS
OUTPUT_DIR    = Path("/kaggle/working/yolov8_output")      # All outputs land here
WEIGHTS_DIR   = OUTPUT_DIR / "weights"                     # Saved model weights
RUNS_DIR      = OUTPUT_DIR / "runs"                        # Ultralytics run logs
EXPORT_DIR    = OUTPUT_DIR / "export"                      # ONNX & compressed zips
# ─────────────────────────────────────────────────────────────────────────────
# CELL 2.1 — Folder Structure Check
# ─────────────────────────────────────────────────────────────────────────────

SUPPORTED_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tiff", ".webp"}

def check_folder_structure(root: Path) -> Dict[str, Path]:
    """
    Validate that the dataset follows YOLO directory conventions.
    Returns a dict of split → (images_dir, labels_dir) tuples.
    """
    print("\n🔍 Checking folder structure...")
    splits = {"train": None, "valid": None, "test": None}
    found = {}

    for split in splits:
        print(split)
        img_dir = root / split / "images"
        lbl_dir = root / split / "labels"

        img_ok = img_dir.exists()
        lbl_ok = lbl_dir.exists()

        status = "✅" if (img_ok and lbl_ok) else ("⚠️ " if img_ok else "❌")
        n_imgs = len([f for f in img_dir.rglob("*") if f.suffix.lower() in SUPPORTED_EXTS]) if img_ok else 0
        n_lbls = len(list(lbl_dir.glob("*.txt"))) if lbl_ok else 0

        print(f"   {status} {split:6s} — images: {n_imgs:5d}  labels: {n_lbls:5d}")

        if img_ok:
            found[split] = {"images": img_dir, "labels": lbl_dir}

    if not found:
        raise FileNotFoundError(
            f"❌ No valid splits found under {root}.\n"
            "   Expected: {root}/images/train/, {root}/images/val/"
        )

    # Also accept flat structure (images/ labels/ without split subdirs)
    flat_img = root / "images"
    flat_lbl = root / "labels"
    if flat_img.exists() and not (flat_img / "train").exists():
        print("   ℹ️  Flat structure detected (no train/val subdirs). You may need to split manually.")

    return found

SPLITS = check_folder_structure(DATASET_ROOT)
# Create output directories
for d in [OUTPUT_DIR, WEIGHTS_DIR, RUNS_DIR, EXPORT_DIR]:
    d.mkdir(parents=True, exist_ok=True)

# ── Training Hyperparameters (centralised) ──────────────────────────────────
# Adjust these BEFORE running training. See Section 4 for detailed explanations.
CFG = dict(
    model_variant  = "yolov8n",   # yolov8n / yolov8s / yolov8m / yolov8l / yolov8x
    img_size       = 640,         # Input resolution (must be multiple of 32)
    batch_size     = 16,          # Reduce to 8 if you see CUDA OOM errors
    epochs         = 100,         # Total training epochs
    workers        = 4,           # DataLoader worker threads (2–4 on Kaggle)
    patience       = 20,          # Early-stopping patience (epochs without improvement)
    lr0            = 0.01,        # Initial learning rate
    lrf            = 0.01,        # Final learning rate = lr0 * lrf
    momentum       = 0.937,       # SGD momentum / Adam beta1
    weight_decay   = 0.0005,      # L2 regularisation
    conf_threshold = 0.25,        # Confidence threshold for inference
    iou_threshold  = 0.45,        # NMS IoU threshold
    project_name   = "yolov8_kaggle",
    run_name       = "baseline",
)

print("✅ Path configuration complete")
print(f"   Dataset  : {DATASET_ROOT}")
print(f"   Output   : {OUTPUT_DIR}")
print(f"\n📋 Training config:")
for k, v in CFG.items():
    print(f"   {k:<16}: {v}")
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 2.2 — Image & Label Integrity Checks
# ─────────────────────────────────────────────────────────────────────────────

def validate_images_and_labels(splits: dict) -> pd.DataFrame:
    """
    For every image in every split:
      - Check the image is readable by OpenCV
      - Check a corresponding label file exists
      - Validate every bounding box in the label file

    Returns a summary DataFrame.
    """
    print("\n🔍 Validating images and labels (this may take a moment)...")
    records = []

    for split, dirs in splits.items():
        img_dir = dirs["images"]
        lbl_dir = dirs["labels"]

        images = [f for f in img_dir.rglob("*") if f.suffix.lower() in SUPPORTED_EXTS]

        bad_images, missing_labels, bad_boxes = 0, 0, 0

        for img_path in images:
            # ── 1. Decode image ───────────────────────────────────────────────
            img = cv2.imread(str(img_path))
            if img is None:
                bad_images += 1
                print(f"   ❌ Corrupt image: {img_path.name}")
                continue

            # ── 2. Find matching label ────────────────────────────────────────
            lbl_path = lbl_dir / (img_path.stem + ".txt")
            if not lbl_path.exists():
                missing_labels += 1
                continue

            # ── 3. Validate bounding boxes ────────────────────────────────────
            with open(lbl_path) as f:
                for line_no, line in enumerate(f, 1):
                    line = line.strip()
                    if not line:
                        continue
                    parts = line.split()
                    if len(parts) != 5:
                        bad_boxes += 1
                        print(f"   ⚠️  Bad format at {lbl_path.name}:{line_no} — expected 5 values, got {len(parts)}")
                        continue
                    try:
                        cls_id, cx, cy, w, h = int(parts[0]), *map(float, parts[1:])
                        # All coords must be in [0, 1] and box must be non-degenerate
                        if not (0 <= cx <= 1 and 0 <= cy <= 1 and 0 < w <= 1 and 0 < h <= 1):
                            bad_boxes += 1
                            print(f"   ⚠️  Out-of-range bbox at {lbl_path.name}:{line_no} → {parts[1:]}")
                    except ValueError:
                        bad_boxes += 1
                        print(f"   ⚠️  Non-numeric values at {lbl_path.name}:{line_no}")

        total = len(images)
        ok = total - bad_images - missing_labels
        records.append({
            "split": split, "total_images": total, "valid_images": ok,
            "corrupt_images": bad_images, "missing_labels": missing_labels,
            "bad_boxes": bad_boxes
        })
        print(f"   {split:6s} — {total} images | {bad_images} corrupt | {missing_labels} missing labels | {bad_boxes} bad boxes")

    df = pd.DataFrame(records).set_index("split")
    print("\n✅ Validation complete")
    return df

validation_summary = validate_images_and_labels(SPLITS)
print("\n", validation_summary.to_string())
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 2.3 — Class Distribution Analysis
# ─────────────────────────────────────────────────────────────────────────────
# Class imbalance is one of the most common causes of poor detection performance.
# This cell counts every class occurrence across the training set.

def count_class_distribution(lbl_dir: Path) -> Counter:
    """Count occurrences of each class ID in a labels directory."""
    counter = Counter()
    for lbl_file in lbl_dir.glob("*.txt"):
        with open(lbl_file) as f:
            for line in f:
                line = line.strip()
                if line:
                    cls_id = int(line.split()[0])
                    counter[cls_id] += 1
    return counter

# Analyse training set distribution
if "train" in SPLITS:
    class_counts = count_class_distribution(SPLITS["train"]["labels"])
    print("\n📊 Class distribution (training set):")
    print(f"   {'Class ID':<12} {'Count':>8} {'Fraction':>10}")
    total_boxes = sum(class_counts.values())
    for cls_id, cnt in sorted(class_counts.items()):
        bar = "█" * int(30 * cnt / max(class_counts.values()))
        print(f"   {cls_id:<12} {cnt:>8} {cnt/total_boxes:>9.1%}  {bar}")

    # Warn on severe imbalance
    min_count = min(class_counts.values())
    max_count = max(class_counts.values())
    ratio = max_count / max(min_count, 1)
    if ratio > 10:
        print(f"\n⚠️  Severe class imbalance detected (ratio {ratio:.0f}:1).")
        print("   Consider: oversampling rare classes, weighted loss, or augmenting rare classes.")

    # Plot
    fig, ax = plt.subplots(figsize=(10, 4))
    ids = sorted(class_counts.keys())
    ax.bar(ids, [class_counts[i] for i in ids], color="steelblue", edgecolor="white")
    ax.set_xlabel("Class ID", fontsize=12)
    ax.set_ylabel("Instance Count", fontsize=12)
    ax.set_title("Class Distribution — Training Set", fontsize=14)
    ax.set_xticks(ids)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "class_distribution.png", dpi=150)
    plt.show()
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 2.4 — Visualize Random Labeled Samples
# ─────────────────────────────────────────────────────────────────────────────
# Visual inspection is irreplaceable. This cell renders N random training images
# with their ground-truth bounding boxes drawn on top.

# ✏️  Update this dict AFTER you know your class names (or load from data.yaml)
CLASS_NAMES: Dict[int, str] = {}   # e.g. {0: "cat", 1: "dog", 2: "car"}

PALETTE = plt.cm.get_cmap("tab20").colors   # 20 distinct colours for boxes

def draw_yolo_boxes(image: np.ndarray, label_path: Path,
                    class_names: dict) -> np.ndarray:
    """Draw YOLO bounding boxes on an image (in-place copy)."""
    img = image.copy()
    h, w = img.shape[:2]

    if not label_path.exists():
        return img   # No labels → return image unchanged

    with open(label_path) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            parts = line.split()
            cls_id = int(parts[0])
            cx, cy, bw, bh = map(float, parts[1:])

            # Convert normalised YOLO coords → pixel coords
            x1 = int((cx - bw / 2) * w)
            y1 = int((cy - bh / 2) * h)
            x2 = int((cx + bw / 2) * w)
            y2 = int((cy + bh / 2) * h)

            color = tuple(int(c * 255) for c in PALETTE[cls_id % len(PALETTE)])
            cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)

            label = class_names.get(cls_id, f"cls_{cls_id}")
            cv2.putText(img, label, (x1, max(y1 - 6, 0)),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)
    return img


def visualize_samples(img_dir: Path, lbl_dir: Path, class_names: dict,
                       n: int = 12) -> None:
    """Display a grid of n random labeled samples."""
    images = [f for f in img_dir.rglob("*") if f.suffix.lower() in SUPPORTED_EXTS]
    samples = random.sample(images, min(n, len(images)))

    cols = 4
    rows = (len(samples) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 4, rows * 3.5))
    axes = axes.flatten()

    for ax, img_path in zip(axes, samples):
        img = cv2.imread(str(img_path))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        lbl_path = lbl_dir / (img_path.stem + ".txt")
        img_drawn = draw_yolo_boxes(img, lbl_path, class_names)
        ax.imshow(img_drawn)
        ax.set_title(img_path.name[:25], fontsize=8)
        ax.axis("off")

    for ax in axes[len(samples):]:
        ax.axis("off")

    plt.suptitle("Random Training Samples with Ground-Truth Boxes", fontsize=14, y=1.01)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "sample_visualization.png", dpi=150, bbox_inches="tight")
    plt.show()


if "train" in SPLITS:
    visualize_samples(
        SPLITS["train"]["images"],
        SPLITS["train"]["labels"],
        CLASS_NAMES
    )
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 3.1 — Generate or Validate data.yaml
# ─────────────────────────────────────────────────────────────────────────────
# ✏️  If you already know your class names, update CLASS_NAMES_LIST here.
# Otherwise, leave it empty and the notebook will use generic names (class_0, ...)

CLASS_NAMES_LIST: List[str] = []   # e.g. ["cat", "dog", "car"]

YAML_OUTPUT_PATH = OUTPUT_DIR / "data.yaml"

def generate_or_validate_yaml(
    dataset_root: Path,
    splits: dict,
    class_names_list: List[str],
    yaml_out: Path
) -> Path:
    """
    If data.yaml already exists in the dataset, load and validate it.
    Otherwise generate one from scratch.

    Returns the path to the validated / generated yaml file.
    """
    existing_yaml = dataset_root / "data.yaml"

    if existing_yaml.exists():
        print("📄 Found existing data.yaml — loading and validating...")
        with open(existing_yaml) as f:
            cfg = yaml.safe_load(f)

        # Validate required keys
        required = {"nc", "names"}
        missing = required - set(cfg.keys())
        if missing:
            raise KeyError(f"data.yaml is missing required keys: {missing}")

        if len(cfg["names"]) != cfg["nc"]:
            raise ValueError(
                f"nc={cfg['nc']} but len(names)={len(cfg['names'])} — they must match."
            )

        # Ensure paths are absolute
        for split_key in ["train", "val", "test"]:
            if split_key in cfg:
                p = Path(cfg[split_key])
                if not p.is_absolute():
                    cfg[split_key] = str(dataset_root / p)

        with open(yaml_out, "w") as f:
            yaml.dump(cfg, f, default_flow_style=False)

        print(f"   ✅ Validated — nc={cfg['nc']}, classes: {cfg['names']}")

    else:
        print("📄 No data.yaml found — generating one...")

        # Infer number of classes from label files
        if "train" in splits:
            all_ids = set()
            for lbl in splits["train"]["labels"].glob("*.txt"):
                with open(lbl) as f:
                    for line in f:
                        if line.strip():
                            all_ids.add(int(line.split()[0]))
            nc = max(all_ids) + 1  # Assumes 0-indexed contiguous class IDs
        else:
            nc = len(class_names_list) if class_names_list else 1

        # Build class name list
        if class_names_list:
            names = class_names_list
        else:
            names = [f"class_{i}" for i in range(nc)]
            print(f"   ⚠️  No class names provided — using generic names. Update CLASS_NAMES_LIST.")

        cfg = {
            # Absolute paths to image directories for each split
            "train": str(splits["train"]["images"]) if "train" in splits else None,
            "val":   str(splits["val"]["images"])   if "val"   in splits else None,
            "test":  str(splits["test"]["images"])  if "test"  in splits else None,

            # Number of object classes — MUST match len(names)
            "nc": nc,

            # Class names list — order matters! class_id 0 → names[0]
            "names": names,
        }
        # Remove None split entries
        cfg = {k: v for k, v in cfg.items() if v is not None}

        with open(yaml_out, "w") as f:
            yaml.dump(cfg, f, default_flow_style=False)

        print(f"   ✅ Generated — nc={nc}, classes: {names}")

    print(f"   📝 YAML saved to: {yaml_out}")
    return yaml_out


YAML_PATH = generate_or_validate_yaml(
    DATASET_ROOT, SPLITS, CLASS_NAMES_LIST, YAML_OUTPUT_PATH
)

# Print the final YAML so you can inspect it
print("\n📄 Final data.yaml contents:")
with open(YAML_PATH) as f:
    print(f.read())

# Load nc and names into Python variables for later use
with open(YAML_PATH) as f:
    _yaml_data = yaml.safe_load(f)
NUM_CLASSES  = _yaml_data["nc"]
CLASS_NAMES  = {i: n for i, n in enumerate(_yaml_data["names"])}
print(f"✅ NUM_CLASSES = {NUM_CLASSES}")
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 4.1 — Load Pretrained YOLOv8 Weights
# ─────────────────────────────────────────────────────────────────────────────
# We always start from COCO-pretrained weights (transfer learning).
# This gives ~5–15 mAP free on most custom datasets vs training from scratch.
# The .pt file is auto-downloaded on first run (~6–22 MB depending on variant).

MODEL_WEIGHTS = f"{CFG['model_variant']}.pt"   # e.g. "yolov8s.pt"

def load_model(weights: str) -> YOLO:
    """Load YOLOv8 with pretrained COCO weights."""
    print(f"\n🏗️  Loading YOLOv8 model: {weights}")
    model = YOLO(weights)

    # Print model info
    n_params = sum(p.numel() for p in model.model.parameters())
    n_train  = sum(p.numel() for p in model.model.parameters() if p.requires_grad)
    print(f"   Parameters (total)     : {n_params:,}")
    print(f"   Parameters (trainable) : {n_train:,}")
    print(f"   Model type             : {type(model.model).__name__}")
    return model

model = load_model(MODEL_WEIGHTS)
print("\n✅ Model ready for training")
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 6.1 — Kaggle Session Keep-Alive
# ─────────────────────────────────────────────────────────────────────────────
# Kaggle sessions can time out after 3 hours of inactivity.
# YOLOv8 training keeps the GPU busy so this is rarely an issue,
# but we log periodic heartbeats to the console to be safe.

import threading

class HeartbeatThread(threading.Thread):
    """Print a heartbeat every N seconds to prevent idle-timeout."""
    def __init__(self, interval_seconds: int = 600):
        super().__init__(daemon=True)   # daemon=True → thread dies when main exits
        self.interval = interval_seconds
        self._stop_event = threading.Event()

    def run(self):
        while not self._stop_event.wait(self.interval):
            elapsed = time.strftime("%H:%M:%S", time.gmtime(time.time() - _START_TIME))
            print(f"💓 Heartbeat — Training in progress | Elapsed: {elapsed}")

    def stop(self):
        self._stop_event.set()

_START_TIME = time.time()
heartbeat = HeartbeatThread(interval_seconds=600)
heartbeat.start()
print("💓 Heartbeat thread started (pings every 10 min)")
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 6.2 — Full Training Run
# ─────────────────────────────────────────────────────────────────────────────

# ── Resume support ──────────────────────────────────────────────────────────
# If your Kaggle session crashed mid-training, find the last checkpoint:
#   /kaggle/working/yolov8_output/runs/<project>/<run>/weights/last.pt
# Then uncomment the line below and run this cell again.
#
# RESUME_FROM = Path("/kaggle/working/yolov8_output/runs/yolov8_kaggle/baseline/weights/last.pt")
RESUME_FROM = None   # Set to a Path to resume, or leave None for fresh run

# ── Training arguments ──────────────────────────────────────────────────────
train_args = dict(
    # ── Dataset ──────────────────────────────────────────────────────────
    data    = str("/kaggle/input/datasets/nikolasgegenava/sard-search-and-rescue/search-and-rescue/data.yaml"),          # Path to data.yaml

    # ── Core hyperparameters ─────────────────────────────────────────────
    epochs  = CFG["epochs"],           # Total training epochs
    imgsz   = CFG["img_size"],         # Input image size (px)
    batch   = CFG["batch_size"],       # Batch size per GPU
    workers = CFG["workers"],          # DataLoader worker processes

    # ── Optimizer ────────────────────────────────────────────────────────
    optimizer    = "SGD",              # "SGD" (default) or "Adam" / "AdamW"
    lr0          = CFG["lr0"],         # Initial learning rate
    lrf          = CFG["lrf"],         # Final lr = lr0 * lrf (cosine decay)
    momentum     = CFG["momentum"],    # SGD momentum
    weight_decay = CFG["weight_decay"],# L2 regularisation coefficient
    warmup_epochs = 3,                 # Linear LR warmup for first N epochs
    warmup_momentum = 0.8,             # Initial momentum during warmup

    # ── Regularisation / stopping ────────────────────────────────────────
    patience   = CFG["patience"],      # Early stopping — 0 to disable
    dropout    = 0.0,                  # Dropout rate (0 = off; try 0.1–0.3)
    label_smoothing = 0.0,             # Label smoothing (0.1 can help noisy labels)

    # ── Precision / performance ──────────────────────────────────────────
    amp     = True,                    # Mixed precision (FP16) — keep True on GPU
    device  = 0,                       # GPU index (0 = first GPU; "cpu" for CPU)
    cache   = False,                   # Cache images in RAM (True if RAM > 16 GB)

    # ── Loss weights ─────────────────────────────────────────────────────
    # These scale the contribution of each loss component.
    # Increase box loss weight for precise localisation tasks.
    # Increase cls loss weight if you have many similar-looking classes.
    box = 7.5,   # Bounding box regression loss weight
    cls = 0.5,   # Classification loss weight
    dfl = 1.5,   # Distribution focal loss weight

    # ── Checkpointing & logging ──────────────────────────────────────────
    project  = str(RUNS_DIR),          # Root output directory for runs
    name     = CFG["run_name"],        # Sub-directory name for this run
    exist_ok = True,                   # Overwrite existing run directory
    save     = True,                   # Save best and last checkpoints
    save_period = 10,                  # Also save every N epochs (0 = off)
    plots    = True,                   # Generate training plots automatically
    verbose  = True,                   # Print per-epoch metrics

    # ── Augmentation (imported from Section 5) ────────────────────────────
    **AUG_CFG,

    # ── Inference thresholds (used for val during training) ───────────────
    conf = CFG["conf_threshold"],
    iou  = CFG["iou_threshold"],
)

# ── Resume logic ────────────────────────────────────────────────────────────
if RESUME_FROM is not None and RESUME_FROM.exists():
    print(f"♻️  Resuming from checkpoint: {RESUME_FROM}")
    model = YOLO(str(RESUME_FROM))
    train_args["resume"] = True
else:
    print(f"🚀 Starting fresh training run: {CFG['run_name']}")
    train_args["resume"] = False

# ── Launch training ──────────────────────────────────────────────────────────
print(f"\n📋 Training configuration summary:")
key_params = ["data", "epochs", "imgsz", "batch", "lr0", "optimizer", "amp", "patience"]
for k in key_params:
    print(f"   {k:<16}: {train_args[k]}")

print("\n⏳ Training starting...")
t0 = time.time()

results = model.train(**train_args)

t1 = time.time()
heartbeat.stop()
print(f"\n✅ Training complete in {(t1-t0)/60:.1f} minutes")
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 6.3 — Copy Best Weights to Safe Location
# ─────────────────────────────────────────────────────────────────────────────
# Ultralytics saves best.pt inside the run directory, but it's easier to
# reference a single canonical path for the best model.

def save_best_weights(results, weights_dir: Path) -> Path:
    """Locate and copy the best weights from the training run."""
    run_dir = Path(results.save_dir)
    best_src = run_dir / "weights" / "best.pt"
    last_src = run_dir / "weights" / "last.pt"

    if best_src.exists():
        best_dst = weights_dir / "best.pt"
        shutil.copy2(best_src, best_dst)
        print(f"✅ Best weights saved: {best_dst}")
        return best_dst
    elif last_src.exists():
        print("⚠️  best.pt not found — using last.pt (training may have been interrupted)")
        dst = weights_dir / "last.pt"
        shutil.copy2(last_src, dst)
        return dst
    else:
        raise FileNotFoundError(f"No weights found in {run_dir / 'weights'}")

BEST_WEIGHTS = save_best_weights(results, WEIGHTS_DIR)
print(f"   Weights path: {BEST_WEIGHTS}")
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 7.2 — Training Loss Curves
# ─────────────────────────────────────────────────────────────────────────────
# Ultralytics saves results.csv in the run directory. We load it and plot
# loss curves to diagnose overfitting / underfitting.

def plot_training_curves(results) -> None:
    """Plot training and validation loss + mAP curves from results.csv."""
    csv_path = Path(results.save_dir) / "results.csv"
    if not csv_path.exists():
        print("⚠️  results.csv not found — skipping loss curves.")
        return

    df = pd.read_csv(csv_path)
    df.columns = df.columns.str.strip()   # Remove whitespace from column names

    fig, axes = plt.subplots(2, 3, figsize=(16, 9))
    axes = axes.flatten()

    # Columns of interest (Ultralytics standard naming)
    plots = [
        ("train/box_loss",   "val/box_loss",   "Box Loss"),
        ("train/cls_loss",   "val/cls_loss",   "Class Loss"),
        ("train/dfl_loss",   "val/dfl_loss",   "DFL Loss"),
        ("metrics/precision(B)", None,          "Precision"),
        ("metrics/recall(B)",    None,          "Recall"),
        ("metrics/mAP50(B)",     "metrics/mAP50-95(B)", "mAP"),
    ]

    for ax, (train_col, val_col, title) in zip(axes, plots):
        if train_col in df.columns:
            ax.plot(df["epoch"], df[train_col], label="Train", color="royalblue")
        if val_col and val_col in df.columns:
            ax.plot(df["epoch"], df[val_col], label="Val", color="tomato", linestyle="--")
        elif train_col in df.columns and val_col is None:
            pass   # Single curve plot
        ax.set_title(title, fontsize=12)
        ax.set_xlabel("Epoch")
        ax.legend()
        ax.grid(True, alpha=0.3)

    plt.suptitle("Training Curves", fontsize=16, y=1.01)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "training_curves.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("✅ Training curves saved.")

    # ── Overfit / underfit diagnosis ─────────────────────────────────────
    if "val/box_loss" in df.columns and "train/box_loss" in df.columns:
        final_train = df["train/box_loss"].iloc[-1]
        final_val   = df["val/box_loss"].iloc[-1]
        gap = final_val - final_train
        if gap > 0.05:
            print(f"\n⚠️  Overfitting detected (val_loss - train_loss = {gap:.3f}).")
            print("   Suggestions: more augmentation, dropout, weight decay, smaller model.")
        elif final_train > 0.5:
            print(f"\n⚠️  Underfitting detected (high train loss = {final_train:.3f}).")
            print("   Suggestions: train longer, larger model, lower weight decay, check data quality.")
        else:
            print(f"\n✅ Loss convergence looks healthy.")

plot_training_curves(results)
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 7.3 — Per-Class mAP Analysis
# ─────────────────────────────────────────────────────────────────────────────

def plot_per_class_map(val_results, class_names: dict) -> None:
    """Bar chart of per-class mAP@50 to identify weak classes."""
    try:
        # ap_class_index contains the class indices evaluated
        ap50_per_class = val_results.box.ap50           # shape: (nc,)
        class_indices  = val_results.box.ap_class_index  # shape: (nc,)
    except AttributeError:
        print("⚠️  Per-class AP data not available in this Ultralytics version.")
        return

    names = [class_names.get(int(i), f"cls_{i}") for i in class_indices]

    fig, ax = plt.subplots(figsize=(max(8, len(names) * 0.7 + 2), 5))
    bars = ax.bar(names, ap50_per_class, color="steelblue", edgecolor="white")

    # Color bars below 0.5 in red
    for bar, val in zip(bars, ap50_per_class):
        if val < 0.5:
            bar.set_color("tomato")

    ax.axhline(y=ap50_per_class.mean(), color="orange", linestyle="--",
               label=f"Mean mAP50 = {ap50_per_class.mean():.3f}")
    ax.set_xlabel("Class", fontsize=12)
    ax.set_ylabel("mAP@50", fontsize=12)
    ax.set_title("Per-Class mAP@50 — Validation Set", fontsize=14)
    ax.set_ylim(0, 1.05)
    ax.legend()
    ax.tick_params(axis="x", rotation=45)

    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "per_class_map.png", dpi=150)
    plt.show()

    # Print weak classes
    weak = [(names[i], float(ap50_per_class[i])) for i in range(len(names))
            if ap50_per_class[i] < 0.5]
    if weak:
        print("\n⚠️  Weak classes (mAP50 < 0.5) — consider:")
        for cls, ap in sorted(weak, key=lambda x: x[1]):
            print(f"   {cls:<20} mAP50 = {ap:.3f}")
        print("   → Collect more samples, adjust augmentation, or check label quality.")

plot_per_class_map(val_results, CLASS_NAMES)
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 7.4 — Display Auto-Generated Evaluation Plots
# ─────────────────────────────────────────────────────────────────────────────
# Ultralytics automatically generates: confusion_matrix.png, PR_curve.png,
# F1_curve.png, P_curve.png, R_curve.png. We display them here.

def display_eval_plots(val_results) -> None:
    """Display Ultralytics-generated evaluation plots."""
    val_dir = Path(val_results.save_dir)
    plot_files = [
        ("confusion_matrix.png", "Confusion Matrix"),
        ("confusion_matrix_normalized.png", "Confusion Matrix (Normalised)"),
        ("PR_curve.png", "Precision-Recall Curve"),
        ("F1_curve.png", "F1 vs Confidence Curve"),
    ]

    found = [(f, t) for f, t in plot_files if (val_dir / f).exists()]
    if not found:
        print("⚠️  No evaluation plots found. Ensure plots=True was set during validation.")
        return

    cols = 2
    rows = (len(found) + 1) // 2
    fig, axes = plt.subplots(rows, cols, figsize=(14, rows * 6))
    axes = axes.flatten()

    for ax, (fname, title) in zip(axes, found):
        img = cv2.imread(str(val_dir / fname))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        ax.imshow(img)
        ax.set_title(title, fontsize=13)
        ax.axis("off")

    for ax in axes[len(found):]:
        ax.axis("off")

    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "eval_plots_combined.png", dpi=150, bbox_inches="tight")
    plt.show()

display_eval_plots(val_results)
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 8.1 — Run Inference on Validation Images
# ─────────────────────────────────────────────────────────────────────────────

def run_inference_and_display(
    model: YOLO,
    img_dir: Path,
    lbl_dir: Path,
    class_names: dict,
    conf: float = 0.25,
    n_samples: int = 8
) -> None:
    """
    Run inference on n_samples random validation images.
    Display ground-truth vs predictions side by side.
    """
    images = [f for f in img_dir.rglob("*") if f.suffix.lower() in SUPPORTED_EXTS]
    samples = random.sample(images, min(n_samples, len(images)))

    fig, axes = plt.subplots(len(samples), 2, figsize=(14, len(samples) * 4))
    if len(samples) == 1:
        axes = [axes]   # Ensure iterable

    for (ax_gt, ax_pred), img_path in zip(axes, samples):
        img_bgr = cv2.imread(str(img_path))
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

        # ── Ground-truth ─────────────────────────────────────────────────
        lbl_path = lbl_dir / (img_path.stem + ".txt")
        gt_img = draw_yolo_boxes(img_rgb, lbl_path, class_names)
        ax_gt.imshow(gt_img)
        ax_gt.set_title("Ground Truth", fontsize=10, color="green")
        ax_gt.axis("off")

        # ── Prediction ───────────────────────────────────────────────────
        result = model.predict(str(img_path), conf=conf, verbose=False)[0]
        pred_img = result.plot(conf=True, line_width=2, font_size=10)
        pred_img_rgb = cv2.cvtColor(pred_img, cv2.COLOR_BGR2RGB)
        ax_pred.imshow(pred_img_rgb)
        n_dets = len(result.boxes)
        ax_pred.set_title(f"Predictions ({n_dets} detections | conf≥{conf})",
                          fontsize=10, color="royalblue")
        ax_pred.axis("off")

    plt.suptitle("Ground Truth vs Predictions", fontsize=14, y=1.01)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "inference_samples.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("✅ Inference visualization complete.")


if "valid" in SPLITS:
    run_inference_and_display(
        best_model,
        SPLITS["valid"]["images"],
        SPLITS["valid"]["labels"],
        CLASS_NAMES,
        conf=CFG["conf_threshold"],
        n_samples=8
    )
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 9.1 — Export to ONNX
# ─────────────────────────────────────────────────────────────────────────────

def export_to_onnx(model: YOLO, export_dir: Path, img_size: int) -> Optional[Path]:
    """Export the YOLOv8 model to ONNX format with dynamic batch axis."""
    print("\n📦 Exporting model to ONNX...")
    try:
        onnx_path = model.export(
            format   = "onnx",
            imgsz    = img_size,
            dynamic  = True,        # Dynamic batch axis — needed for variable-batch inference
            simplify = True,        # Run ONNX simplifier (removes redundant ops)
            opset    = 12,          # ONNX opset — 12 is widely compatible
        )
        # Move to export dir
        dst = export_dir / "best.onnx"
        shutil.copy2(onnx_path, dst)
        size_mb = dst.stat().st_size / 1e6
        print(f"   ✅ ONNX model saved: {dst} ({size_mb:.1f} MB)")
        return dst
    except Exception as e:
        print(f"   ❌ ONNX export failed: {e}")
        return None

ONNX_PATH = export_to_onnx(best_model, EXPORT_DIR, CFG["img_size"])
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# CELL 9.2 — Package Outputs for Download
# ─────────────────────────────────────────────────────────────────────────────
# Kaggle working directory (/kaggle/working) is persistent across sessions.
# We create a single compressed archive for easy download.

def package_outputs(output_dir: Path, weights_dir: Path,
                    export_dir: Path) -> Path:
    """Zip key outputs into a single downloadable archive."""
    zip_path = output_dir / "yolov8_results.zip"
    print(f"\n📦 Packaging outputs → {zip_path}")

    with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
        # Weights
        for pt in weights_dir.glob("*.pt"):
            zf.write(pt, arcname=f"weights/{pt.name}")
            print(f"   + weights/{pt.name}")

        # ONNX
        for onnx in export_dir.glob("*.onnx"):
            zf.write(onnx, arcname=f"export/{onnx.name}")
            print(f"   + export/{onnx.name}")

        # Metrics
        metrics_path = output_dir / "val_metrics.json"
        if metrics_path.exists():
            zf.write(metrics_path, arcname="val_metrics.json")
            print(f"   + val_metrics.json")

        # YAML
        yaml_path = output_dir / "data.yaml"
        if yaml_path.exists():
            zf.write(yaml_path, arcname="data.yaml")
            print(f"   + data.yaml")

        # Plots
        for png in output_dir.glob("*.png"):
            zf.write(png, arcname=f"plots/{png.name}")
            print(f"   + plots/{png.name}")

    size_mb = zip_path.stat().st_size / 1e6
    print(f"\n✅ Archive created: {zip_path.name} ({size_mb:.1f} MB)")
    print(f"   → Download from the Kaggle Output tab on the right panel")
    return zip_path

ARCHIVE_PATH = package_outputs(OUTPUT_DIR, WEIGHTS_DIR, EXPORT_DIR)
```
